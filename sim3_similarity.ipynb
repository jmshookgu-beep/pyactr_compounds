Everything below uses the functions defined in [`compound_parser.ipynb`](https://github.com/jmshookgu-beep/pyactr_compounds/blob/main/compound_parser.ipynb). Long-term memory is set up the same way in every simulation: word chunks get frequency-based base levels (wordfreq), compound links get base levels from SemEval-2013 paraphrase proportions, `finalize_memory()` adds 1,000 background links with Tratz & Hovy (2010) relation proportions and sets S = ln(number of chunks in memory).

Two novel compounds, *snack money* and *cocaine money*, have no link chunks of their own. With partial matching, they are interpreted through the familiar *money* compound whose modifier is most similar (GloVe): SNACK is close to DINNER, so *snack money* gets FOR; COCAINE is close to DRUG, so *cocaine money* gets the dominant relation of *drug money*, FROM.


```python
def show_links(dm, names):
    for name, ch in names.items():
        print(f"{name:14s} B = {base_level(dm, ch):.3f}")

glove = None
def setup():
    global glove
    parser, environment, dm = build_parser(partial_matching=True)
    glove = set_glove_similarities(parser, glove=glove)   # downloads glove-wiki-gigaword-100 (~128 MB) on first run
    links = {"DRUG FOR":  add_background_link(dm, "MONEY", "DRUG",  "FOR",  p=26/66),
             "DRUG FROM": add_background_link(dm, "MONEY", "DRUG",  "FROM", p=36/66),
             "DINNER FOR": add_background_link(dm, "MONEY", "DINNER", "FOR", p=1.0)}
    finalize_memory(parser, dm)
    show_links(dm, links)
    return parser, environment, dm
```

Similarity scores used for partial matching:

```python
parser, environment, dm = setup()
for a, b in [("snack", "dinner"), ("snack", "drug"), ("cocaine", "dinner"), ("cocaine", "drug")]:
    cos = float(glove.similarity(a, b))
    print(f"{a:7s} - {b:6s} cosine = {cos:.3f}   match score = {max(0.0, cos) - 1:.3f}")
```

## snack money

```python
parser, environment, dm = setup()
run(parser, environment, "the snack money is safe")
```
