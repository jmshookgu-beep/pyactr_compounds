Everything below uses the functions defined in [`compound_parser.ipynb`](https://github.com/jmshookgu-beep/pyactr_compounds/blob/main/compound_parser.ipynb). Long-term memory is set up the same way in every simulation: word chunks get frequency-based base levels (wordfreq), compound links get base levels from SemEval-2013 paraphrase proportions, `finalize_memory()` adds 1,000 background links with Tratz & Hovy (2010) relation proportions and sets S = ln(number of chunks in memory).

*the drug money is safe*. The two MONEY–DRUG links carry their long-term history from SemEval-2013 (FROM 36/66: *money made from drugs*; FOR 26/66: *money to buy drugs*), so the more frequent relation, FROM, is retrieved. In the recency variant, the FOR link has also been used every 10 s over the last minute, which is enough to make FOR win.

```python
def show_links(dm, names):
    for name, ch in names.items():
        print(f"{name:14s} B = {base_level(dm, ch):.3f}")

def setup(recency=False):
    parser, environment, dm = build_parser()
    links = {"DRUG FOR":  add_background_link(dm, "MONEY", "DRUG", "FOR",  p=26/66),
             "DRUG FROM": add_background_link(dm, "MONEY", "DRUG", "FROM", p=36/66)}
    if recency:
        add_link(dm, "MONEY", "DRUG", "FOR", every=60, since=60)
    finalize_memory(parser, dm)
    show_links(dm, links)
    return parser, environment, dm
```

## default: long-term history only

```python
parser, environment, dm = setup()
run(parser, environment, "the drug money is safe")
```

## recency variant: FOR used once 60 seconds prior


```python
parser, environment, dm = setup(recency=True)
run(parser, environment, "the drug money is safe")
```
