*a stencil design won*. The two link chunks for STENCIL–DESIGN carry their long-term usage history (SemEval-2013 proportions: MADE .69, OF .18). The more frequent relation, MADE, should be retrieved.

```python
parser, environment, dm = build_parser()
# long-term link histories from SemEval-2013 Task 4 paraphrases
# (N = 1000 encounters per compound, lifespan L = 20 years; see add_background_link)
STENCIL_OF   = add_background_link(dm, "DESIGN", "STENCIL", "OF",   p=0.183)
STENCIL_MADE = add_background_link(dm, "DESIGN", "STENCIL", "MADE", p=0.690)

# recency variant: uncomment to give OF a recent history (used every 10 s over the last minute)
# add_link(dm, "DESIGN", "STENCIL", "OF", every=10, since=60)

for name, ch in {"STENCIL MADE": STENCIL_MADE, "STENCIL OF": STENCIL_OF}.items():
    print(f"{name:14s} B = {base_level(dm, ch):.3f}")
```
