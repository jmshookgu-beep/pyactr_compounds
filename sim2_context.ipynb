Everything below uses the functions defined in [`compound_parser.ipynb`](https://github.com/jmshookgu-beep/pyactr_compounds/blob/main/compound_parser.ipynb). Long-term memory is set up the same way in every simulation: word chunks get frequency-based base levels (wordfreq), compound links get base levels from SemEval-2013 paraphrase proportions, `finalize_memory()` adds 1,000 background links with Tratz & Hovy (2010) relation proportions and sets S = ln(number of chunks in memory).

*drug money* is ambiguous in SemEval-2013: FROM (*money made from drugs*, 36/66) vs FOR (*money to buy drugs*, 26/66). On its own it gets the more frequent FROM reading. After *dinner money* (FOR: *money to buy dinner*; not in SemEval, so treated as unambiguous), the imaginal buffer still holds the relation FOR, which spreads activation to the FOR link of *drug money* and makes it win.

```python
def show_links(dm, names):
    for name, ch in names.items():
        print(f"{name:14s} B = {base_level(dm, ch):.3f}")

def setup():
    parser, environment, dm = build_parser()
    links = {"DRUG FROM": add_background_link(dm, "MONEY", "DRUG",  "FROM", p=36/66),
             "DRUG FOR":  add_background_link(dm, "MONEY", "DRUG",  "FOR",  p=26/66),
             "DINNER FOR": add_background_link(dm, "MONEY", "DINNER", "FOR", p=1.0)}
    finalize_memory(parser, dm)
    show_links(dm, links)
    return parser, environment, dm
```

## primed by *dinner money*

```python
parser, environment, dm = setup()
run(parser, environment, "the dinner money is gone but the drug money is safe")
```
