*the road design is done but the stencil design is late*. *Road design* is mostly OF (.82). After it is parsed, the imaginal buffer still holds its link chunk (relation OF), which spreads activation to the OF link of *stencil design*.

```python
parser, environment, dm = build_parser()
# long-term link histories from SemEval-2013 Task 4 paraphrases
# (N = 1000 encounters per compound, lifespan L = 20 years; see add_background_link)
STENCIL_MADE = add_background_link(dm, "DESIGN", "STENCIL", "MADE", p=0.690)
STENCIL_OF   = add_background_link(dm, "DESIGN", "STENCIL", "OF",   p=0.183)
ROAD_OF      = add_background_link(dm, "DESIGN", "ROAD",    "OF",   p=0.824)

for name, ch in {"STENCIL MADE": STENCIL_MADE, "STENCIL OF": STENCIL_OF, "ROAD OF": ROAD_OF}.items():
    print(f"{name:14s} B = {base_level(dm, ch):.3f}")
```

```python
run(parser, environment, "the road design is done but the stencil design is late")
```
